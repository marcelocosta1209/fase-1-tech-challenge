```python
# ============================================================
# CARREGAR MODELOS TREINADOS NOS NOTEBOOKS ANTERIORES
# ============================================================

modelo_tabular = joblib.load('../models/melhor_modelo_tabular.pkl')
scaler = joblib.load('../models/scaler.pkl')
print(f'Modelo tabular carregado: {type(modelo_tabular).__name__}')

cancer = load_breast_cancer()
feature_names = cancer.feature_names
print(f'Features: {len(feature_names)}')

# Tentar carregar modelo CNN
modelo_cnn = None
CNN_DISPONIVEL = False
if TF_DISPONIVEL:
    try:
        modelo_cnn = keras.models.load_model('../models/modelo_cnn_mama.keras')
        CNN_DISPONIVEL = True
        print('Modelo CNN carregado com sucesso!')
    except Exception as e:
        print(f'Modelo CNN nao encontrado: {e}')
else:
    print('TensorFlow nao disponivel. Analise de imagem sera conceitual.')
```

# PARTE 1: CASO CLÍNICO - DADOS TABULARES (FNA)

## 3. Dados do Caso Clínico

**Paciente:** Mulher, 52 anos  
**Queixa:** Nódulo palpável na mama direita detectado em exame de rotina  
**Exame:** Aspiração por Agulha Fina (FNA)  

Os valores abaixo são medições computadorizadas das células coletadas pela FNA:

```python
# ============================================================
# SIMULAR DADOS DE UM CASO CLÍNICO REAL
# ============================================================
# Usamos um caso real do dataset (amostra que o modelo NUNCA viu)
# Isso simula a situação de um novo paciente chegando ao hospital

# Selecionar um caso maligno do dataset para demonstração
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df['diagnostico'] = cancer.target

# Pegar um caso maligno (diagnóstico = 0)
caso_maligno = df[df['diagnostico'] == 0].iloc[15]  # Caso #15 dos malignos
caso_benigno = df[df['diagnostico'] == 1].iloc[20]  # Para comparação

# Exibir dados do caso principal
print('='*70)
print('CASO CLÍNICO #1 - PACIENTE COM NÓDULO SUSPEITO')
print('='*70)
print('\nDados da Aspiração por Agulha Fina (FNA):')
print('-'*70)

# Organizar dados em formato legível
dados_caso = pd.DataFrame({
    'Feature': feature_names,
    'Valor Medido': caso_maligno[feature_names].values.round(4)
})

# Adicionar valores de referência (média dos benignos vs malignos)
dados_caso['Média Benigno'] = df[df['diagnostico']==1][feature_names].mean().values.round(4)
dados_caso['Média Maligno'] = df[df['diagnostico']==0][feature_names].mean().values.round(4)

# Indicar se o valor está mais próximo de benigno ou maligno
dados_caso['Tendência'] = dados_caso.apply(
    lambda row: 'MALIGNO' if abs(row['Valor Medido'] - row['Média Maligno']) < abs(row['Valor Medido'] - row['Média Benigno']) else 'Benigno',
    axis=1
)

# Mostrar as 10 features mais relevantes
print('\nTop 10 Features mais relevantes do caso:')
features_top10 = ['worst concave points', 'worst perimeter', 'mean concave points',
                   'worst radius', 'mean perimeter', 'worst area',
                   'mean radius', 'mean area', 'mean concavity', 'worst concavity']
dados_caso[dados_caso['Feature'].isin(features_top10)]
```

## 4. Predição do Modelo Tabular

```python
# ============================================================
# FAZER A PREDIÇÃO PARA O CASO CLÍNICO
# ============================================================

# Preparar os dados do caso (mesmo pipeline do treino)
X_caso = caso_maligno[feature_names].values.reshape(1, -1)
X_caso_df = pd.DataFrame(X_caso, columns=feature_names)

# Aplicar o scaler (MESMO que foi usado no treino)
X_caso_scaled = scaler.transform(X_caso_df)
X_caso_scaled_df = pd.DataFrame(X_caso_scaled, columns=feature_names)

# Fazer predição
predicao = modelo_tabular.predict(X_caso_scaled_df)[0]
probabilidades = modelo_tabular.predict_proba(X_caso_scaled_df)[0]

# Exibir resultado
classe = 'MALIGNO' if predicao == 0 else 'BENIGNO'
prob_maligno = probabilidades[0]  # Probabilidade da classe 0 (maligno)
prob_benigno = probabilidades[1]  # Probabilidade da classe 1 (benigno)

print('='*70)
print('RESULTADO DA ANÁLISE - MODELO DE MACHINE LEARNING')
print('='*70)
print(f'\n  Classificação: {classe}')
print(f'  Probabilidade de Maligno: {prob_maligno:.1%}')
print(f'  Probabilidade de Benigno: {prob_benigno:.1%}')
print(f'\n  Confiança do modelo: {max(prob_maligno, prob_benigno):.1%}')

# Classificação de risco
if prob_maligno > 0.8:
    risco = 'ALTO RISCO - Altamente suspeito de malignidade'
    recomendacao = 'Biópsia URGENTE recomendada'
elif prob_maligno > 0.5:
    risco = 'RISCO MODERADO - Suspeito'
    recomendacao = 'Biópsia recomendada + exames complementares'
elif prob_maligno > 0.2:
    risco = 'RISCO BAIXO - Indeterminado'
    recomendacao = 'Acompanhamento em 6 meses + exames complementares'
else:
    risco = 'RISCO MUITO BAIXO - Provavelmente benigno'
    recomendacao = 'Acompanhamento de rotina'

print(f'\n  Classificação de Risco: {risco}')
print(f'  Recomendação: {recomendacao}')
print(f'\n  ⚠ ATENÇÃO: Este resultado é apenas uma ferramenta de APOIO.')
print(f'  O diagnóstico definitivo deve ser feito pelo médico.')
```

```python
# ============================================================
# VISUALIZAÇÃO DA PROBABILIDADE
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico de barras com probabilidades
cores = ['#e74c3c', '#2ecc71']
barras = axes[0].bar(['Maligno', 'Benigno'], [prob_maligno, prob_benigno],
                      color=cores, edgecolor='black', alpha=0.8)
axes[0].set_title('Probabilidades do Modelo', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Probabilidade')
axes[0].set_ylim(0, 1.1)
# Adicionar valor sobre cada barra
for barra, valor in zip(barras, [prob_maligno, prob_benigno]):
    axes[0].text(barra.get_x() + barra.get_width()/2, barra.get_height() + 0.02,
                f'{valor:.1%}', ha='center', fontweight='bold', fontsize=14)
axes[0].axhline(y=0.5, color='gray', linestyle='--', alpha=0.5, label='Limiar 50%')
axes[0].legend()

# Gauge/termômetro de risco
# Criar uma barra de risco horizontal
risk_colors = ['#2ecc71', '#f1c40f', '#e67e22', '#e74c3c']
risk_labels = ['Baixo', 'Moderado', 'Alto', 'Muito Alto']
risk_ranges = [0.2, 0.5, 0.8, 1.0]

# Desenhar a escala de risco
start = 0
for color, label, end in zip(risk_colors, risk_labels, risk_ranges):
    axes[1].barh(0, end - start, left=start, height=0.5, color=color, edgecolor='black', alpha=0.7)
    axes[1].text((start + end)/2, 0, label, ha='center', va='center', fontsize=9, fontweight='bold')
    start = end

# Marcar a posição do paciente
axes[1].axvline(x=prob_maligno, color='black', linewidth=3, linestyle='-')
axes[1].plot(prob_maligno, 0, 'v', color='black', markersize=15)
axes[1].text(prob_maligno, 0.35, f'{prob_maligno:.1%}', ha='center', fontsize=12, fontweight='bold')

axes[1].set_xlim(0, 1)
axes[1].set_title('Escala de Risco - Probabilidade de Malignidade', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Probabilidade de Maligno')
axes[1].set_yticks([])

plt.tight_layout()
plt.savefig('../data/caso_clinico_probabilidades.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 5. Explicação com SHAP (Caso Individual)

```python
# ============================================================
# SHAP - EXPLICAÇÃO INDIVIDUAL DO CASO CLÍNICO
# ============================================================

print('Calculando SHAP values para o caso clínico...')

if hasattr(modelo_tabular, 'estimators_'):  # Random Forest
    explainer = shap.TreeExplainer(modelo_tabular)
    shap_values_raw = explainer.shap_values(X_caso_scaled)
    base_val = explainer.expected_value
else:
    df_treino = pd.DataFrame(cancer.data, columns=feature_names)
    X_bg = shap.sample(scaler.transform(df_treino), 50)
    explainer = shap.KernelExplainer(modelo_tabular.predict_proba, X_bg)
    shap_values_raw = explainer.shap_values(X_caso_scaled)
    base_val = explainer.expected_value

# Extrair classe Maligno (classe 0) - tratamento universal
raw = np.array(shap_values_raw)
if isinstance(shap_values_raw, list):
    shap_vals_maligno = np.array(shap_values_raw[0])
elif raw.ndim == 3:
    shap_vals_maligno = raw[:, :, 0]
else:
    shap_vals_maligno = raw

if isinstance(base_val, (list, np.ndarray)):
    base_value = np.array(base_val).flatten()[0]
else:
    base_value = base_val

print(f'SHAP values shape: {shap_vals_maligno.shape}')
print('SHAP values calculados!\n')
```

```python
# ============================================================
# WATERFALL PLOT - Explicação detalhada do caso
# ============================================================
# Mostra como cada feature contribuiu para a predição DESTE paciente

plt.figure(figsize=(12, 10))
shap.waterfall_plot(
    shap.Explanation(
        values=shap_vals_maligno[0],
        base_values=base_value,
        data=X_caso_scaled_df.iloc[0].values,
        feature_names=feature_names.tolist()
    ),
    max_display=15,
    show=False
)
plt.title('SHAP Waterfall - Contribuicao de Cada Feature para o Diagnostico',
          fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('../data/shap_waterfall_caso.png', dpi=150, bbox_inches='tight')
plt.show()

print('Interpretacao do grafico:')
print('  - Cada barra mostra a contribuicao de uma feature')
print('  - Vermelho: empurra para MALIGNO')
print('  - Azul: empurra para BENIGNO')
print('  - O comprimento da barra = magnitude do efeito')
```

```python
# ============================================================
# FORCE PLOT - Visualização compacta do SHAP
# ============================================================
# Mostra as forças que empurram a predição para cada lado

shap.initjs()

plt.figure(figsize=(20, 4))
shap.force_plot(
    base_value,
    shap_vals_maligno[0],
    X_caso_scaled_df.iloc[0].values,
    feature_names=feature_names.tolist(),
    matplotlib=True,
    show=False
)
plt.title('SHAP Force Plot - Caso Clinico', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../data/shap_force_caso.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ============================================================
# ANÁLISE DETALHADA DAS FEATURES MAIS INFLUENTES
# ============================================================
# Identificar quais features mais contribuíram para o diagnóstico

# Criar DataFrame com SHAP values
shap_df = pd.DataFrame({
    'Feature': feature_names,
    'SHAP Value': shap_vals_maligno[0],
    'Valor Original': X_caso[0],
    'Valor Normalizado': X_caso_scaled[0]
})

# Ordenar por impacto absoluto
shap_df['Impacto Absoluto'] = abs(shap_df['SHAP Value'])
shap_df = shap_df.sort_values('Impacto Absoluto', ascending=False)
shap_df['Direção'] = shap_df['SHAP Value'].apply(lambda x: 'MALIGNO' if x > 0 else 'Benigno')

print('='*70)
print('TOP 10 FEATURES QUE MAIS INFLUENCIARAM O DIAGNÓSTICO')
print('='*70)
print('\n(Positivo = empurra para Maligno | Negativo = empurra para Benigno)\n')

for i, row in shap_df.head(10).iterrows():
    direcao = 'MALIGNO' if row['SHAP Value'] > 0 else 'Benigno'
    seta = '>>' if row['SHAP Value'] > 0 else '<<'
    print(f"  {row['Feature']:30s} | SHAP: {row['SHAP Value']:+.4f} | {seta} {direcao}")
```

## 6. Análise de Imagem de Mamografia

```python
# Analise de imagem com Grad-CAM
IMG_SIZE = (224, 224)

if CNN_DISPONIVEL and TF_DISPONIVEL:
    def gerar_gradcam(img_array, modelo, layer_name="Conv_1"):
        last_conv = modelo.get_layer(layer_name)
        grad_model = tf.keras.models.Model(inputs=modelo.input, outputs=[last_conv.output, modelo.output])
        with tf.GradientTape() as tape:
            conv_out, pred = grad_model(img_array)
            loss = pred[:, 0]
        grads = tape.gradient(loss, conv_out)
        weights = tf.reduce_mean(grads, axis=(0, 1, 2))
        heatmap = conv_out[0] @ weights[..., tf.newaxis]
        heatmap = tf.squeeze(heatmap)
        heatmap = tf.maximum(heatmap, 0) / (tf.math.reduce_max(heatmap) + 1e-8)
        return heatmap.numpy()

    test_dir = "../data/images/breast_cancer/test"
    img_path, classe_real = None, None
    for classe in ["malignant", "benign"]:
        cd = os.path.join(test_dir, classe)
        if os.path.exists(cd):
            arqs = [f for f in os.listdir(cd) if f.endswith((".png",".jpg"))]
            if arqs: img_path, classe_real = os.path.join(cd, arqs[0]), classe; break

    if img_path:
        print(f"Analisando: {img_path}")
        img = keras.utils.load_img(img_path, target_size=IMG_SIZE)
        img_arr = keras.utils.img_to_array(img) / 255.0
        img_batch = np.expand_dims(img_arr, 0)
        prob = modelo_cnn.predict(img_batch, verbose=0)[0][0]
        class_names = ["benign", "malignant"]
        pred = class_names[int(prob > 0.5)]
        conf = prob if prob > 0.5 else 1 - prob
        hm = gerar_gradcam(img_batch, modelo_cnn)
        hm_r = cv2.resize(hm, IMG_SIZE)
        hm_c = cv2.applyColorMap(np.uint8(255*hm_r), cv2.COLORMAP_JET)
        hm_c = cv2.cvtColor(hm_c, cv2.COLOR_BGR2RGB) / 255.0
        overlay = np.clip(hm_c * 0.4 + img_arr, 0, 1)
        fig, axes = plt.subplots(1, 3, figsize=(15, 5))
        axes[0].imshow(img_arr); axes[0].set_title(f"Original ({classe_real.upper()})", fontweight="bold"); axes[0].axis("off")
        axes[1].imshow(hm_r, cmap="jet"); axes[1].set_title("Grad-CAM"); axes[1].axis("off")
        axes[2].imshow(overlay)
        mask = hm_r > 0.7; coords = np.where(mask)
        if len(coords[0]) > 0:
            from matplotlib.patches import Rectangle
            axes[2].add_patch(Rectangle((coords[1].min(),coords[0].min()),coords[1].max()-coords[1].min(),coords[0].max()-coords[0].min(),lw=2,ec="yellow",fc="none",ls="--"))
        axes[2].set_title(f"Pred: {pred.upper()} ({conf:.0%})", fontweight="bold", color="green" if pred==classe_real else "red"); axes[2].axis("off")
        plt.suptitle("Analise de Mamografia - Caso Clinico", fontsize=16, fontweight="bold")
        plt.tight_layout()
        plt.savefig("../data/caso_clinico_imagem.png", dpi=150, bbox_inches="tight")
        plt.show()
    else: print("Nenhuma imagem encontrada.")
else:
    print("CNN nao disponivel. Execute Notebook 03 primeiro.")
    print("A analise tabular acima esta completa.")
```

## 9. Discussão Crítica e Análise de Aplicabilidade

### O modelo pode ser utilizado na prática?

**Sim, mas exclusivamente como ferramenta de APOIO ao diagnóstico.** O sistema proposto:

#### Aplicação Prática Proposta:

1. **Triagem inicial**: O sistema analisa automaticamente os resultados do FNA e/ou mamografia, classificando o risco do paciente

2. **Priorização de casos**: Pacientes classificados como alto risco pelo modelo são direcionados para atendimento prioritário

3. **Segunda opinião**: O médico usa o SHAP e Grad-CAM para entender o raciocínio do modelo e comparar com sua avaliação clínica

4. **Redução de tempo**: Acelera a análise inicial, permitindo que o médico foque nos casos mais complexos

#### Por que o Recall é a métrica mais importante?

Em diagnóstico de câncer:
- **Falso Negativo** (modelo diz benigno, mas é maligno) = paciente pode perder a chance de tratamento precoce. **ISSO É INACEITÁVEL.**
- **Falso Positivo** (modelo diz maligno, mas é benigno) = paciente faz exames adicionais desnecessários. **Isso é tolerável.**

Portanto, preferimos um modelo com **alto Recall** (detecta o máximo de malignos possível), mesmo que tenha mais falsos positivos.

#### Limitações do Sistema:

| Limitação | Impacto | Mitigação |
|-----------|---------|----------|
| Dataset limitado (569 amostras tabulares) | Generalização limitada | Validar com dados reais do hospital |
| Não considera histórico do paciente | Pode ignorar fatores de risco | Integrar com prontuário eletrônico |
| Sensível à qualidade dos dados de entrada | Predições incorretas com dados ruins | Protocolo de qualidade na coleta |
| Classificação binária simplificada | Não distingue subtipos de câncer | Expandir para classificação multiclasse |

#### Considerações Éticas:

- **Transparência**: O paciente deve ser informado que IA foi usada no processo diagnóstico
- **Responsabilidade**: A decisão final é SEMPRE do médico
- **Viés**: O modelo foi treinado em dados específicos e pode não representar todas as populações
- **Privacidade**: Dados médicos devem seguir LGPD/HIPAA

---
*O(a) médico(a) SEMPRE deve ter a palavra final no diagnóstico.*
