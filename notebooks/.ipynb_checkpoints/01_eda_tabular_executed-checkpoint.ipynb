## 1. Importação de Bibliotecas

```python
# ============================================================
# IMPORTAÇÃO DAS BIBLIOTECAS NECESSÁRIAS
# ============================================================
import pandas as pd                          # Manipulação e análise de dados tabulares
import numpy as np                           # Operações numéricas e arrays
import matplotlib.pyplot as plt              # Criação de gráficos e visualizações
import seaborn as sns                        # Gráficos estatísticos avançados
from sklearn.datasets import load_breast_cancer  # Dataset de câncer de mama do sklearn

# Configurações de visualização para melhor apresentação dos gráficos
plt.rcParams['figure.figsize'] = (12, 6)    # Tamanho padrão dos gráficos (largura, altura)
plt.rcParams['font.size'] = 12              # Tamanho da fonte nos gráficos
sns.set_style('whitegrid')                   # Estilo visual com grade branca
pd.set_option('display.max_columns', None)   # Mostrar todas as colunas do DataFrame

print('Bibliotecas carregadas com sucesso!')
```

## 2. Carregamento do Dataset

O Breast Cancer Wisconsin é um dataset clássico para classificação binária na área médica. Foi coletado por Dr. William H. Wolberg na Universidade de Wisconsin e contém medições computadorizadas de imagens de aspiração por agulha fina (FNA) de massas mamárias.

```python
# ============================================================
# CARREGAMENTO DO DATASET BREAST CANCER WISCONSIN
# ============================================================
# O dataset já vem incluso no scikit-learn, não precisa baixar CSV
# Contém medições de células tumorais obtidas por FNA (aspiração por agulha fina)

cancer = load_breast_cancer()  # Carregar o dataset do sklearn

# Criar DataFrame pandas para facilitar a manipulação dos dados
# cancer.data = matriz com os valores das 30 features
# cancer.feature_names = nomes das features
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)

# Adicionar a coluna alvo (target): 0 = Maligno, 1 = Benigno
df['diagnostico'] = cancer.target
# Criar coluna com nomes legíveis para uso nos gráficos
df['diagnostico_nome'] = df['diagnostico'].map({0: 'Maligno', 1: 'Benigno'})

print(f'Dataset carregado: {df.shape[0]} amostras, {df.shape[1]} colunas')
print(f'Features: {len(cancer.feature_names)}')
print(f'Classes: {cancer.target_names}')
df.head()  # Exibir as 5 primeiras linhas para inspeção visual
```

## 3. Visão Geral do Dataset

```python
# ============================================================
# VISÃO GERAL DO DATASET - informações estruturais
# ============================================================
print('='*60)
print('INFORMAÇÕES DO DATASET')
print('='*60)
print(f'\nDimensões: {df.shape}')           # (linhas, colunas)
print(f'Total de amostras: {df.shape[0]}')  # Número de pacientes
print(f'Total de features: {df.shape[1] - 2}')  # Exclui diagnostico e diagnostico_nome

# Verificar tipos de dados (todas devem ser numéricas)
print(f'\nTipos de dados:')
print(df.dtypes.value_counts())

# Verificar valores nulos (dados faltantes)
# Valores nulos podem causar erros nos modelos de ML
print(f'\nValores nulos por coluna:')
nulos = df.isnull().sum()
print(f'Total de valores nulos: {nulos.sum()}')
if nulos.sum() > 0:
    print(nulos[nulos > 0])  # Mostrar apenas colunas com nulos
else:
    print('Nenhum valor nulo encontrado!')
```

```python
# ============================================================
# ESTATÍSTICAS DESCRITIVAS
# ============================================================
# .describe() calcula: count, mean, std, min, 25%, 50%, 75%, max
# .T transpõe a tabela para melhor visualização (features nas linhas)
print('ESTATÍSTICAS DESCRITIVAS')
print('='*60)
df.describe().T
```

### Discussão - Visão Geral

O dataset possui **569 amostras** e **30 features numéricas**, sem valores nulos. As features representam 10 características reais dos núcleos celulares, cada uma com 3 variações:
- **mean** (média)
- **error** (erro padrão)
- **worst** (pior valor = média dos 3 maiores)

As 10 características base são: radius, texture, perimeter, area, smoothness, compactness, concavity, concave points, symmetry e fractal dimension.

## 4. Distribuição da Variável Alvo (Diagnóstico)

```python
# ============================================================
# GRÁFICO DE DISTRIBUIÇÃO DA VARIÁVEL ALVO (DIAGNÓSTICO)
# ============================================================
# Verifica se as classes estão balanceadas ou desbalanceadas
# Isso impacta diretamente na escolha do modelo e métricas

# Contar quantas amostras de cada classe existem
contagem = df['diagnostico_nome'].value_counts()
# Calcular a proporção percentual de cada classe
proporcao = df['diagnostico_nome'].value_counts(normalize=True) * 100

print('Distribuição do Diagnóstico:')
print('-'*40)
for classe in contagem.index:
    print(f'{classe}: {contagem[classe]} ({proporcao[classe]:.1f}%)')

# Criar figura com 2 subplots lado a lado
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico de barras - mostra contagem absoluta
colors = ['#e74c3c', '#2ecc71']  # Vermelho para maligno, verde para benigno
axes[0].bar(contagem.index, contagem.values, color=colors, edgecolor='black', alpha=0.8)
axes[0].set_title('Distribuição do Diagnóstico', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Quantidade')
# Adicionar valor numérico sobre cada barra
for i, v in enumerate(contagem.values):
    axes[0].text(i, v + 5, str(v), ha='center', fontweight='bold', fontsize=13)

# Gráfico de pizza - mostra proporção percentual
axes[1].pie(contagem.values, labels=contagem.index, autopct='%1.1f%%',
            colors=colors, startangle=90, explode=(0.05, 0),
            textprops={'fontsize': 13})
axes[1].set_title('Proporção do Diagnóstico', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('../data/distribuicao_diagnostico.png', dpi=150, bbox_inches='tight')
plt.show()

# Calcular razão entre classes (indica nível de desbalanceamento)
print(f'\nRazão Benigno/Maligno: {contagem["Benigno"]/contagem["Maligno"]:.2f}')
```

### Discussão - Variável Alvo

O dataset apresenta um leve **desbalanceamento** com ~63% de casos benignos e ~37% malignos. Isso é importante porque:
- Modelos podem ter viés para a classe majoritária (benigno)
- **Recall** será uma métrica crucial: em diagnóstico de câncer, é pior deixar de detectar um caso maligno (falso negativo) do que gerar um falso alarme (falso positivo)
- O desbalanceamento não é severo o suficiente para exigir técnicas como SMOTE, mas devemos monitorar as métricas por classe

## 5. Histogramas das Features Principais

```python
# ============================================================
# HISTOGRAMAS DAS FEATURES 'MEAN' (médias)
# ============================================================
# Histogramas mostram a distribuição dos valores de cada feature
# Separamos por classe (Maligno vs Benigno) para ver se há diferença
# Se as distribuições se sobrepõem muito = feature pouco discriminativa
# Se as distribuições são bem separadas = feature boa para classificação

# Selecionar apenas features que contêm 'mean' no nome
features_mean = [col for col in df.columns if 'mean' in col]

# Criar grade de subplots: 2 linhas x 5 colunas = 10 gráficos
fig, axes = plt.subplots(2, 5, figsize=(20, 8))
axes = axes.ravel()  # Transformar matriz 2D em array 1D para iterar

for i, feat in enumerate(features_mean):
    # Plotar histograma de cada classe com cores diferentes
    for diag, color, label in [(0, '#e74c3c', 'Maligno'), (1, '#2ecc71', 'Benigno')]:
        axes[i].hist(df[df['diagnostico'] == diag][feat], bins=25,
                     alpha=0.6, color=color, label=label, edgecolor='black', linewidth=0.5)
    axes[i].set_title(feat.replace(' ', '\n'), fontsize=10, fontweight='bold')
    axes[i].legend(fontsize=8)

plt.suptitle('Distribuição das Features (Mean) por Diagnóstico', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('../data/histogramas_features_mean.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ============================================================
# HISTOGRAMAS DAS FEATURES 'WORST' (piores valores)
# ============================================================
# Features 'worst' representam a média dos 3 maiores valores
# Tendem a ser mais discriminativas que 'mean' e 'error'

features_worst = [col for col in df.columns if 'worst' in col]

fig, axes = plt.subplots(2, 5, figsize=(20, 8))
axes = axes.ravel()

for i, feat in enumerate(features_worst):
    for diag, color, label in [(0, '#e74c3c', 'Maligno'), (1, '#2ecc71', 'Benigno')]:
        axes[i].hist(df[df['diagnostico'] == diag][feat], bins=25,
                     alpha=0.6, color=color, label=label, edgecolor='black', linewidth=0.5)
    axes[i].set_title(feat.replace(' ', '\n'), fontsize=10, fontweight='bold')
    axes[i].legend(fontsize=8)

plt.suptitle('Distribuição das Features (Worst) por Diagnóstico', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('../data/histogramas_features_worst.png', dpi=150, bbox_inches='tight')
plt.show()
```

### Discussão - Histogramas

Observamos uma **boa separação** entre as classes em várias features:
- **mean radius, mean perimeter, mean area**: tumores malignos são significativamente maiores
- **mean concavity, mean concave points**: malignos têm maior concavidade (bordas irregulares)
- **smoothness e fractal dimension**: pouca separação entre classes

Isso indica que **tamanho e forma irregular** são os melhores indicadores de malignidade, o que é consistente com a literatura médica.

## 7. Análise de Correlação

```python
# ============================================================
# HEATMAP DE CORRELAÇÃO (features mean + diagnóstico)
# ============================================================
# A correlação de Pearson mede a relação linear entre duas variáveis
# Valor de -1 a +1:
#   +1 = correlação positiva perfeita (quando uma sobe, outra sobe)
#   -1 = correlação negativa perfeita (quando uma sobe, outra desce)
#    0 = sem correlação linear

# Selecionar apenas features mean + diagnóstico para visualização
df_mean = df[features_mean + ['diagnostico']]
corr_mean = df_mean.corr()  # Calcular matriz de correlação

plt.figure(figsize=(14, 10))
# mask: ocultar triângulo superior (informação redundante)
mask = np.triu(np.ones_like(corr_mean, dtype=bool))
# Heatmap com anotação dos valores e colormap divergente (azul-branco-vermelho)
sns.heatmap(corr_mean, annot=True, fmt='.2f', cmap='RdBu_r', center=0,
            mask=mask, square=True, linewidths=0.5,
            cbar_kws={'shrink': 0.8})
plt.title('Matriz de Correlação - Features Mean + Diagnóstico', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../data/correlacao_mean.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ============================================================
# CORRELAÇÃO DE TODAS AS FEATURES COM O DIAGNÓSTICO
# ============================================================
# Mostra quais features têm maior relação com o resultado (maligno/benigno)
# Correlação negativa = quanto maior o valor, mais provável ser MALIGNO
# (pois maligno=0 e benigno=1 no dataset)

# Calcular correlação de cada feature com a coluna 'diagnostico'
corr_diagnostico = df.drop(columns=['diagnostico_nome']).corr()['diagnostico'].drop('diagnostico').sort_values()

plt.figure(figsize=(10, 12))
# Colorir barras: vermelho para correlação negativa (indica malignidade), verde para positiva
colors = ['#e74c3c' if v < 0 else '#2ecc71' for v in corr_diagnostico.values]
corr_diagnostico.plot(kind='barh', color=colors, edgecolor='black', alpha=0.8)
plt.title('Correlação das Features com o Diagnóstico', fontsize=14, fontweight='bold')
plt.xlabel('Coeficiente de Correlação de Pearson')
plt.axvline(x=0, color='black', linewidth=0.8)  # Linha central no zero
plt.tight_layout()
plt.savefig('../data/correlacao_diagnostico.png', dpi=150, bbox_inches='tight')
plt.show()

# Listar as top 10 features mais correlacionadas (em valor absoluto)
print('\nTop 10 features mais correlacionadas (valor absoluto):')
print(corr_diagnostico.abs().sort_values(ascending=False).head(10))
```

```python
# Heatmap completo
plt.figure(figsize=(20, 16))
corr_full = df.drop(columns=['diagnostico_nome']).corr()
mask = np.triu(np.ones_like(corr_full, dtype=bool))
sns.heatmap(corr_full, cmap='RdBu_r', center=0, mask=mask,
            square=True, linewidths=0.3, cbar_kws={'shrink': 0.6},
            xticklabels=True, yticklabels=True)
plt.title('Matriz de Correlação Completa', fontsize=16, fontweight='bold')
plt.xticks(fontsize=7, rotation=90)
plt.yticks(fontsize=7)
plt.tight_layout()
plt.savefig('../data/correlacao_completa.png', dpi=150, bbox_inches='tight')
plt.show()
```

### Discussão - Correlação

Observações importantes:

1. **Alta multicolinearidade** entre features do mesmo grupo:
   - radius, perimeter e area são altamente correlacionadas (~0.99) — faz sentido, pois são medidas de tamanho
   - compactness, concavity e concave points também correlacionam entre si

2. **Features mais correlacionadas com o diagnóstico** (correlação negativa = mais maligno):
   - worst concave points (-0.79)
   - worst perimeter (-0.78)
   - mean concave points (-0.78)
   - worst radius (-0.78)

3. **Implicações para modelagem:**
   - A multicolinearidade pode afetar modelos lineares (como Regressão Logística)
   - Random Forest e KNN são mais robustos a isso
   - Podemos considerar PCA ou seleção de features no pré-processamento

## 8. Pairplot das Features Mais Relevantes

```python
# ============================================================
# PAIRPLOT - RELAÇÃO ENTRE AS 5 FEATURES MAIS CORRELACIONADAS
# ============================================================
# O pairplot mostra gráficos de dispersão entre pares de features
# A diagonal mostra a distribuição (KDE) de cada feature individual
# Permite identificar visualmente clusters e separação de classes

# Selecionar as 5 features com maior correlação absoluta com o diagnóstico
top5_features = corr_diagnostico.abs().sort_values(ascending=False).head(5).index.tolist()
print(f'Top 5 features: {top5_features}')

# Criar subset do DataFrame com apenas essas features + classe
df_pair = df[top5_features + ['diagnostico_nome']]

# Gerar pairplot colorido por classe
g = sns.pairplot(df_pair, hue='diagnostico_nome',
                 palette={'Maligno': '#e74c3c', 'Benigno': '#2ecc71'},
                 diag_kind='kde',              # KDE na diagonal (curva de densidade)
                 plot_kws={'alpha': 0.5, 's': 20})  # Pontos semi-transparentes
g.figure.suptitle('Pairplot - Top 5 Features por Correlação', y=1.02, fontsize=14, fontweight='bold')
plt.savefig('../data/pairplot_top5.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 9. Análise de Outliers

```python
# ============================================================
# DETECÇÃO DE OUTLIERS USANDO O MÉTODO IQR
# ============================================================
# IQR (Interquartile Range) = Q3 - Q1
# Outliers são valores abaixo de Q1 - 1.5*IQR ou acima de Q3 + 1.5*IQR
# Em dados biomédicos, outliers podem ser casos clínicos reais importantes!

print('ANÁLISE DE OUTLIERS (Método IQR)')
print('='*60)

features_numericas = cancer.feature_names.tolist()
outlier_counts = {}

# Para cada feature, calcular os limites e contar outliers
for feat in features_numericas:
    Q1 = df[feat].quantile(0.25)       # Primeiro quartil (25%)
    Q3 = df[feat].quantile(0.75)       # Terceiro quartil (75%)
    IQR = Q3 - Q1                      # Amplitude interquartil
    lower = Q1 - 1.5 * IQR            # Limite inferior
    upper = Q3 + 1.5 * IQR            # Limite superior
    # Contar quantos valores estão fora dos limites
    n_outliers = ((df[feat] < lower) | (df[feat] > upper)).sum()
    if n_outliers > 0:
        outlier_counts[feat] = n_outliers

# Organizar resultados em DataFrame
outlier_df = pd.DataFrame.from_dict(outlier_counts, orient='index', columns=['n_outliers'])
outlier_df = outlier_df.sort_values('n_outliers', ascending=False)
outlier_df['percentual'] = (outlier_df['n_outliers'] / len(df) * 100).round(1)

print(f'\nFeatures com outliers: {len(outlier_df)} de {len(features_numericas)}')
print(f'\nTop 10 features com mais outliers:')
outlier_df.head(10)
```

### Discussão - Outliers

- A maioria das features possui outliers, o que é normal em dados biomédicos
- **Não removeremos os outliers** pois podem representar casos clínicos reais importantes
- Em vez disso, usaremos **StandardScaler** no pré-processamento, que é mais robusto
- Modelos como Random Forest são naturalmente resistentes a outliers
